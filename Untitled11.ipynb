```python
class Node:
    def __init__(self, value=None):
        self.value = value
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def add_node(self, value_dict):
        new_node = Node(value_dict)
        if self.head is None:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node

    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(current.value)
            current = current.next
        for item in elements:
            print(item)
```

```python
my_list = LinkedList()

my_list.add_node({"id": 1, "name": "ETHAN"})
my_list.add_node({"id": 3, "county": "KISUMU"})
my_list.add_node({"id": 2, "name": "ALAINE", "age": 21})

print("Linked List elements:")
my_list.display()
```
